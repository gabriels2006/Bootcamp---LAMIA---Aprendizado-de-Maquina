```python
%matplotlib inline

import numpy as np

incomes = np.random.normal(27000, 15000, 10000)
# gera rendas aleatórias

incomes = np.append(incomes, [1000000000])
# adiciona um outlier gigante, fora do comum

import matplotlib.pyplot as plt

plt.hist(incomes, 50) # cria histograma

plt.show()
```

```python
print(incomes.mean()) #mostrar a media
```

```python
def reject_outliers(data):
    u = np.median(data)
    s = np.std(data) #dp

    filtered = [e for e in data if (u - 2 * s < e < u + 2 * s)]
    # remove valores muito distantes da média dos dados

    return filtered

filtered = reject_outliers(incomes)# aplica o filtro nos salários

plt.hist(filtered, 50)
plt.show()
```

```python
print(np.mean(filtered)) #media filtrada sem o outlier
```
