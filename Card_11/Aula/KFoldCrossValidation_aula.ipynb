```python
#na importação tive que mudar por causa da versão do SKLEARN - deixou de ser cross_validation e passou a ser from sklearn.model_selection import train_test_split para importar a mesma coisa. Portanto, em todo o código da aula vai estar referenciando essa parte, não cross validation.

import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.model_selection import cross_val_score
from sklearn import datasets
from sklearn import svm
#aqui carregamos o dataset já usado na aula de PCA
iris = datasets.load_iris()
```

```python
from sklearn.model_selection import train_test_split
#basicamente o que isso faz é separar os dados do dataset para os treinamentos, separando 40% para os testes
X_train, X_test, y_train, y_test = train_test_split(
    iris.data,
    iris.target,
    test_size=0.4,
    random_state=0
)

#constrói um modelo SVC - Support Vector Classifier -
clf = svm.SVC(kernel='linear', C=1).fit(X_train, y_train)

clf.score(X_test, y_test)
```

```python
scores = cross_val_score( #linear, mas sendo um polinomial poderia ser melhor - aqui separa por linhas
    clf, #modelo
    iris.data, #dados completos
    iris.target, #os valores reais
    cv=5 #aqui ele divide para 5 datasets - folds
)

print(scores)
print(scores.mean())
```

```python
clf = svm.SVC(kernel='poly', C=1).fit(X_train, y_train) #aqui é o polinomial - o teste foi melhor - já que pode usar curvas
scores = cross_val_score(
    clf,
    iris.data,
    iris.target,
    cv=5
)

print(scores)
print(scores.mean()) #mesmo assim, os resultados foram mais satisfatórios do que o da aula, no meu caso. O modelo não deu o overfitting igual ao do professor
```
