# Reinforcement Learning

```python
import gym
import random

random.seed(1234)

streets = gym.make("Taxi-v3",render_mode='ansi').env #nova forma (descricao do card)
streets.reset()
print("\n" + streets.render())
```

```python
initial_state = streets.encode(2, 3, 2, 0)

streets.s = initial_state #aqui colocamos o estado inicial, no caso, na frente do r.

print("\n" + streets.render())
```

```python
streets.P[initial_state] #aqui tratamos de probabilidade, como nada foi treinado, tudo está com 1 - máximo.
```

```python
import numpy as np
import random

q_table = np.zeros([streets.observation_space.n, streets.action_space.n]) #array 2d que representa todos os estados

learning_rate = 0.1 #taxa de aprendizado
discount_factor = 0.6 # o quanto o agente valoriza recompensas futuras
exploration = 0.1 #taxa de exploracao
epochs = 10000 #total de vezes que roda para treinar

for taxi_run in range(epochs):

    # Gymnasium retorna (state, info), então na nova versão pega apenas o estado
    state = streets.reset()[0]
    done = False

    while not done:
        random_value = random.uniform(0, 1)
        if random_value < exploration:
            action = streets.action_space.sample()  # Explore a random action - andar por ai
        else:
            action = np.argmax(q_table[state])  # Use the action with the highest q-value - memoriza o melhor valor pra seguir ele

        # Gymnasium retorna 5 valores abaixo
        next_state, reward, terminated, truncated, info = streets.step(action)

        # O episódio termina se for terminated ou truncated
        done = terminated or truncated

        prev_q = q_table[state, action]
        next_max_q = np.max(q_table[next_state])
        new_q = (1 - learning_rate) * prev_q + learning_rate * (
            reward + discount_factor * next_max_q
        )

        q_table[state, action] = new_q

        state = next_state
```

```python
from IPython.display import clear_output
from time import sleep
#nessa etapa treinamos o modelo com reforco e estamos usando esse treinamento.
for tripnum in range(1, 11):
    state = streets.reset()[0]

    done = False

    while not done:
        action = np.argmax(q_table[state])

        next_state, reward, terminated, truncated, info = streets.step(action)
        done = terminated or truncated

        clear_output(wait=True)
        print("Trip number " + str(tripnum))
        print(streets.render())
        sleep(.5)

        state = next_state

    sleep(.1) #aqui deixei mais rápido para poder enxergar, 2 segundos era demais, muito parado.
```
