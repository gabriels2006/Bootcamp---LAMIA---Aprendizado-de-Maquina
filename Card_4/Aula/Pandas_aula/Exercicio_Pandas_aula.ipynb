```python
def chefe_cargo(cargo):
    if 'chief' in cargo.lower():
        return True
    else:
        return False #palavra chefe em seu cargo precisa da função, contabilizando isso
```

```python
sum(sal['JobTitle'].apply(lambda x: chefe_cargo(x))) # lambda como função rápida
```

```python
sal['tam'] = sal['JobTitle'].apply(len) #existe correlação entre o comprimento da sequencia do titulo do trabalho e o salario? - não.
```

```python
sal[['tam', 'TotalPayBenefits']].corr() #correlaciona as duas colunas
```
