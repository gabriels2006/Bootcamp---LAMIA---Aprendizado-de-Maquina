DataFrame - Criação e Fatiamento

```python
dados = {
    'Nome': ['Ana', 'Bruno', 'Carlos', 'Daniel'],
    'Idade': [23, 35, 45, 29],
    'Cidade': ['SP', 'RJ', 'BH', 'POA'],
    'Nota': [8.5, 7.0, 9.2, 6.8]
} #dicionário base
```

```python
df = pd.DataFrame(dados) #associa df a um dataframe do conjunto de dados
```

```python
df
```

```python
df[['Nome','Nota']] #filtro por nome e nota
```

```python
df.head(2) #apenas as 2 primeiras linhas
```

Seleção condicional e set_index

```python
df[df['Idade'] > 30] #nomes com 31 anos pra cima
```

```python
df_indexado = df.set_index('Nome')
```

```python
df_indexado
```

GroupBy

```python
df_group = df.groupby('Cidade')['Nota'].mean() #agrupando cidade e nota
```

```python
df_group
```

Concatenar, Juntar e Mesclar


```python
df2 = pd.DataFrame({'Nome':['Ana','Carlos'], 'Salario':[5000,7000]})
```

```python
df_merge = pd.merge(df, df2, on='Nome', how='left')
```

```python
df_merge #junção dos dados
```

```python
df_concat = pd.concat([df, df_merge], axis=0)
```

```python
df_concat #valores relacionados concatenados
```
