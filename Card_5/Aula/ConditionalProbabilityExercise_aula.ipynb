```python
from numpy import random
random.seed(0)
#esse bloco todo vai simular a probabilidade de compra por tal faixa etária, poderia ser aleatório, mas temos que ter dados consistentes pra esse exemplo
totals = {20:0, 30:0, 40:0, 50:0, 60:0, 70:0}
purchases = {20:0, 30:0, 40:0, 50:0, 60:0, 70:0}
totalPurchases = 0

for _ in range(100000):
    ageDecade = random.choice([20, 30, 40, 50, 60, 70])
    purchaseProbability = float(ageDecade) / 100.0
    totals[ageDecade] += 1
    if (random.random() < purchaseProbability):
        totalPurchases += 1
        purchases[ageDecade] += 1
```

```python
totals #valores espalhados (totais individuais)
```

```python
purchases #valores das compras
```

```python
totalPurchases #somatório geral
```

```python
PEF = float(purchases[30]/float(totals[30]))  #Probabilidade de comparar aos 30 anos/a compraram
print('P(purschases | 30s): ' + str(PEF))
```

```python
PF = float(totals[30]/100000.0)
print("P(30's): " + str(PF)) #probabilidade de ter 30 no data set
```

```python
PE = float(totalPurchases)/100000.0
print("P(30's): " + str(PE))  #independente da idade
```

```python
print("P(30's, Purchase)" + str(float(purchases[30])/100000.0)) #fora da população
```

```python
print("P(30's)P(Purchase)" + str(PE*PF)) # é basicamente a mesma coisa que se colocasse P(E,F) na estatistica, funciona da mesma forma que na matematica
```

```python
print((purchases[30] / 100000.0) / PF)
```
