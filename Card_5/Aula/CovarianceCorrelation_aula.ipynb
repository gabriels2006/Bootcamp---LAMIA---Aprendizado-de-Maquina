```python
import numpy
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
from matplotlib.pyplot import scatter


#funções que definem covarianca e media, fazendo automático e aleatório com np.random printando a funcão
def de_mean(x):
    xmean = np.mean(x)
    return [xi - xmean for xi in x]

def covariance(x,y):
    n = len(x)
    return np.dot(de_mean(x), de_mean(y)) / (n-1)

pageSpeeds = np.random.normal(3.0, 1.0, 1000)
purchaseAmount = np.random.normal(50.0, 10.0, 1000)

plt.scatter(pageSpeeds, purchaseAmount)

covariance(pageSpeeds, purchaseAmount)
```

```python
purchaseAmount = np.random.normal(5.0, 10.0, 1000)
plt.scatter(pageSpeeds, purchaseAmount) #novamente, normal
covariance(pageSpeeds, purchaseAmount)
```

```python
purchaseAmount = np.random.normal(5.0, 10.0, 1000) / pageSpeeds
plt.scatter(pageSpeeds, purchaseAmount) #dividido pela page speeds, aparece o gráfico "linear" das compras
covariance(pageSpeeds, purchaseAmount)
```

```python
def correlation(x,y):
    stddevx = x.std()
    stddevy = y.std()
    return covariance(x, y) / stddevx / stddevx #aqui seria o equivalente ao desvio padrão, entre -1 e 1
correlation(pageSpeeds, purchaseAmount)
```

```python
np.corrcoef(pageSpeeds, purchaseAmount) #numpy faz automatico com essa função (mais simples de usar)
```

```python
purchaseAmount = 100 - pageSpeeds * 3
plt.scatter(pageSpeeds, purchaseAmount)
correlation(pageSpeeds, purchaseAmount) #forçãondo matematicamente uma correlação mais certa e linear, melhorando a visualização
```
