```python
import numpy as np
import matplotlib.pyplot as plt

values = np.random.uniform(-10.0, 10.0, 100000) #criando o gráfico em barras plano/uniforme

plt.hist(values, 50)
plt.show()
```

```python
from scipy.stats import norm
import matplotlib.pyplot as plt

x = np.arange(-3, 3, 0.001) #dados aleatórios
plt.plot(x, norm.pdf(x)) #gráfico 2d para a função gaussiana/normal para densidade
```

```python
import numpy as np
import matplotlib.pyplot as plt

mu = 5.0
sigma = 2.0
values = np.random.normal(mu, sigma, 10000) #distribuição com o sigma, ou seja, padrão.
plt.hist(values, 50)
plt.show()
```

```python
from scipy.stats import expon
import matplotlib.pyplot as plt

x = np.arange(0, 10 , 0.001)
plt.plot(x, expon.pdf(x)) #função exponencial (limite a 0)
```

```python
from scipy.stats import binom
import matplotlib.pyplot as plt

n, p = 10, 0.5
x = np.arange(0, 10, 0.001)
plt.plot(x, binom.pmf(x, n, p)) #função binomial
```

```python
from scipy.stats import poisson
import matplotlib.pyplot as plt

mu = 500
x = np.arange(400, 600, 0.5)
plt.plot(x, poisson.pmf(x, mu))
```
