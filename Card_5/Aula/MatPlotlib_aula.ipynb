```python
import matplotlib.pyplot as plt
import numpy as np

# Ativa o estilo xkcd só dentro do bloco pois estava crashnado todos os graficos nesse arquivo
with plt.xkcd():
    # Força uma fonte já instalada no meu sistema
    plt.rcParams['font.family'] = 'Arial'

    fig = plt.figure() #cria a figura (1x1x1 tamanho)
    ax = fig.add_subplot(1, 1, 1)
    ax.spines['right'].set_color('none') #sem cor nas linhas de borda
    ax.spines['top'].set_color('none')
    plt.xticks([])
    plt.yticks([])
    ax.set_ylim([-30, 10]) #tamanho novamente

    data = np.ones(100) # cria um vetor de 1
    data[70:] -= np.arange(30) #aleatóriio

    plt.annotate( #o que vai ser printado e onde aparece
       'THE DAY I REALIZED\nI COULD COOK BACON\nWHENEVER I WANTED',
        xy=(70, 1), arrowprops=dict(arrowstyle='->'), xytext=(15, -10))

    plt.plot(data)

    plt.xlabel('time')
    plt.ylabel('my overall health')
    plt.show()
```

```python
# Remove XRCD mode:

plt.rcdefaults()

values = [12, 55, 4, 32, 14] #porcentagem, cores, o quanto fica pra fora, Nomes, dados e titulo
colors = ['r', 'g', 'b', 'c', 'm']
explode = [0, 0, 0.2, 0, 0]
labels = ['India', 'United States', 'Russia', 'China', 'Europe']
plt.pie(values, colors = colors, labels=labels, explode=explode)
plt.title('Student Locations')
plt.show()
```

```python
values = [12, 55, 4, 32, 14]
colors = ['r', 'g', 'b', 'c', 'm']
plt.bar(range(0,5), values, color = colors) #grafico de barras colorida - Valores, cores, altura
plt.show()
```

```python
# Dados uniformes entre -40 e 60
uniformSkewed = np.random.rand(100) * 100 - 40
# outliers altos (acima de 100) e outliers baixos (abaixo de -100)
high_outliers = np.random.rand(10) * 50 + 100
low_outliers = np.random.rand(10) * -50 - 100
data = np.concatenate((uniformSkewed, high_outliers, low_outliers)) #juncção em um array
plt.boxplot(data)
plt.show()
```
