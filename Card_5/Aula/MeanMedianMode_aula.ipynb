```python
from random import randint

import numpy as np
```

```python
incomes = np.random.normal(27000, 15000, 10000)  #distribuição aleatória
```

```python
np.mean(incomes) #printando os incomes, média
```

```python
%matplotlib inline
import matplotlib.pyplot as plt
plt.hist(incomes, 50) #cria um histograma com 50 barras
plt.show()
```

```python
np.median(incomes)
```

```python
incomes = np.append(incomes, [1000000000]) #adicionando o "Jeff Bezos", um bilhão
```

```python
np.mean(incomes) #media
```

```python
np.median(incomes) #mediana
```
