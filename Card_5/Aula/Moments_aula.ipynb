```python
import numpy as np
import matplotlib.pyplot as plt

vals = np.random.normal(0, 0.5, 10000) #gráfico em barra com dados aleatórios

plt.hist(vals, 50)
plt.show()
```

```python
np.mean(vals) #media
```

```python
np.var(vals) #variação
```

```python
import scipy.stats as sp
sp.skew(vals) #distribuição normal
```

```python
sp.kurtosis(vals) #mede o tamanho da calda
```
