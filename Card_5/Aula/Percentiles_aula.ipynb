```python
import numpy as np
import matplotlib.pyplot as plt

vals = np.random.normal(0, 0.5, 10000) #gráfico em barra com dados aleatórios

plt.hist(vals, 50)
plt.show()
```

```python
np.percentile(vals, 50) #conseidera 50 porcento (seria a mediana)
```

```python
np.percentile(vals, 90) #considerando 90 porcento
```

```python
np.percentile(vals, 20) #20%
```
