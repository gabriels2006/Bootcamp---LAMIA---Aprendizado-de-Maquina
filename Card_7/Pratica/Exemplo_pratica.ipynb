Gráfico simples para exemplo de dados no colab

```python
import matplotlib.pyplot as plt

carros = ['Gol', 'Celta', 'Uno', 'Palio', 'Ka']
vendas = [120, 95, 80, 70, 60]

fig, ax = plt.subplots()

ax.bar(carros, vendas)

ax.set(
    xlabel='Modelos',
    ylabel='Quantidade Vendida',
    title='Vendas de Carros'
)

ax.grid(axis='y')

plt.show()
```
