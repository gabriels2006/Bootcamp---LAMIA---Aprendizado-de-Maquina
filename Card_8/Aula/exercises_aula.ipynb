```python
files = webpage.select("div.block a")
relative_files = [file['href'] for file in files]

url = "https://keithgalli.github.io/web-scraping/"
for f in relative_files:
    full_url = url + f
    page = requests.get(full_url)
    bs_page = bs(page.content)
    secret_word_element = bs_page.find("p", attrs={"id": "secret-word"})
    secret_word = secret_word_element.string
    print(secret_word) # a piada pronta, ter certeza de dar o like. Aqui aparece completo e a chave secreta finalmente é desvendada. O que acontece aqui é como um filtro
```
