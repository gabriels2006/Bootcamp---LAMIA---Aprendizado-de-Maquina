## Load in the necessary libraries

```python
import requests #pip install requests
from bs4 import BeautifulSoup as bs #pip install beautifulsoup4
```

Load our first page

```python
#Load the webpage content - carregar a página no r
r = requests.get("https://keithgalli.github.io/web-scraping/example.html")

#Convert to a beautiful soup object
soup = bs(r.content)

#Print ou our HTML - imprimir o html
# print(soup)
print(soup.prettify()) #para melhorar a legibilidade
```

Start using Beautiful Soup to Scrape

Find e Find All

```python
first_header = soup.find("h2")
first_header #mostra o primeiro elemento

headers = soup.find_all("h2")
print(headers) #mostra os 2 headers
```

```python
# Pass in a list of elements to look for - passar uma lista gera para obsrvar
first_header = soup.find(["h1", "h2"])
first_header #independente da ordem, o resultado é o mesmo

headers = soup.find_all(["h1", "h2"])
headers
```

```python
#you can pass in attributes to the fin/find_all function
paragraph = soup.find_all("p")
paragraph #retorna todos os p no html

paragraph = soup.find_all("p", attrs={"id": "paragraph-id"})
paragraph #retorna todos os p singulares
```

```python
#You can nest find/findall calls
body = soup.find('body')
body #aqui mostra o corpo do html
```

```python
body = soup.find('body')
div = body.find('div')
header = div.find('h1')
header #basicamente aqui é feita a comparação e transformada para header do div, abaixo apenas melhora a visualização
```

```python
# We can seanch specific strings in our find/findall calls
print(soup.prettify())
```

```python
# We can search specific strings in our find/find_all calls - Acha escritas especificas dentro do html
paragraphs = soup.find_all("p", string="Some")
paragraphs #o retorno é nulo aqui
```

```python
import re
paragraphs = soup.find_all("p", string=re.compile("Some"))
paragraphs #o retorno aqui é certo

headers = soup.find_all("h2", string=re.compile("(H|h)eader"))
headers #voltamos a ter um uico pois só h'um, mas colocando (H|h) acha todos os dois.
```

# Select CSS selector

```python
#no site que ele mostra do w3schols tem muitas funções do css.
content = soup.select("p")
content
```

```python
print(soup.body.prettify())
```

```python
content = soup.select("div p")
content #mostrar apenas paragrafos dentro da div
```

```python
paragraphs = soup.select("h2 ~ p")
paragraphs #aqui vem os que tem um h2 e depois é seguido por um paragrafo
```

```python
#mostrar o elemento em negrito do paragrafo
bold_text = soup.select("p#paragraph-id b")
bold_text
```

```python
paragraphs = soup.select("body > p")
paragraphs

for paragraph in paragraphs:
    print(paragraph.select("i")) #nao tem elementos em italico
```

```python
# Grab by element with specific property
soup.select("[align=middle]")
```

# Get different properties of the HTML

```python
#Quer o elemento dentro do header
header = soup.find("h2")
header.string

div = soup.find("div")
print(div.prettify())
print(div.string) #nao é claro para a funçao onde
```

```python
#Quer o elemento dentro do header
header = soup.find("h2")
header.string

div = soup.find("div")
print(div.prettify())
print(div.get_text())#agora pega o texto em multiplos elementos
```

```python
# Get a specific property from an element - propriedade do elemento
link = soup.find("a")
link['href']

paragraphs = soup.select("p#paragraph-id")
paragraphs[0]['id']
```
