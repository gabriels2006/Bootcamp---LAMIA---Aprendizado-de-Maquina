```python
import requests
from bs4 import BeautifulSoup as bs
import re
```

```python
url = "https://www.youtube.com/watch?v=-22HuKQgMEI"
```

```python
r = requests.get(url, headers={'User-Agent': 'Mozilla/5.0'}) # carregando a página no r

#Convertendo para visualizar melhor
scrap = bs(r.content)

#output
print(scrap.prettify())
```

```python
first_header =  scrap.find("h2") #achando as tags h2 no html do youtube
first_header

headers = scrap.find_all("h2")
print(headers) # o outpyt é vazio pois o yt não tem cabecalho assim, mas tem titulo e imagens que vou explorar depois.
```

```python
div = scrap.find("div")

if div:
    header = div.find("h1")
    print(header)
else:
    print("Nenhuma div encontrada")
```

```python
header = div.find_all('h1')
print(header)
#O retorno é nulo pois o youtube está codificado principalmente em JavaScipt.
```

```python
#Ao invés de persistir em procurar uma H1, as meta tags são boas referências, elas agem como as h1.

title = scrap.find(
    "meta",
    attrs={"property": "og:title"}
)
#Aqui busco apenas o titulo
print(title)
print(title["content"])
```

```python
#Já aqui, o objetivo é baixar a imagem, assim como na aula, mas sendo a thumbnail do vídeo
thumbnail = scrap.find(
    "meta",
    attrs={"property": "og:image"}
)

print(thumbnail["content"])

#consegui o link da imagem, basta agora baixar
```

```python
#mas como um aprendiz em ciencia de dados, nao vai ser na mao. Passo a url e com o get pego a imagem, depois baixo :)

url = "https://i.ytimg.com/vi/-22HuKQgMEI/maxresdefault.jpg"

img_data = requests.get(url).content

with open("thumbnail_youtube_aula.jpg", "wb") as handler:
    handler.write(img_data)

print("Imagem salva, deu certo")
```

```python
meta_tags = scrap.find_all("meta") #tudo o que for meta tag vai ir automatico
len(meta_tags)

#numero de meta tags que eu posso explorar
```

```python
for tag in meta_tags:
    content = tag.get("content")

    if content and re.search("(C|c)hance", content, re.I):
        print(content)

#Aqui busca todas as palavras Chance, com c maiusculo e minusculo, apresentado na aula
```
